```python

ticker = yf.Ticker("AAPL")

current_price = ticker.fast_info["lastPrice"]
print(current_price)
```

```python
info = ticker.fast_info

print(info["lastPrice"])      # Current price
print(info["dayHigh"])        # Highest price today
print(info["dayLow"])         # Lowest price today
print(info["open"])           # Today's opening price
print(info["previousClose"])  # Yesterday's closing price
print(info["lastVolume"])     # Today's trading volume
print(info["marketCap"])      # Company's market value
```

```python
import yfinance as yf

ticker = yf.Ticker("AAPL")

history = ticker.history(period="1mo")
print(history)
info = ticker.info

print(info["currentPrice"])
print(info["previousClose"])
print(info["open"])
print(info["dayHigh"])
print(info["dayLow"])
```

```python
# Download Apple historical data
aapl = yf.download("AAPL", period="max")

# Find highest price
highest_price = aapl["High"].max()

# Find the date of highest price
highest_date = aapl["High"].idxmax()
```
